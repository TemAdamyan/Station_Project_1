Online Payments Fraud Detection

1. import the data

```python
import pandas as pd

data = pd.read_csv('PS.csv')
data
```

1.1 visualize like pie

```python
import plotly.express as px

fig = px.pie(data, 
             names=data['isFraud'].map({0: 'Not Fraud', 1: 'Fraud'}), 
             title='Fraudulent vs Non-Fraudulent Transactions', 
             color_discrete_sequence=px.colors.sequential.RdBu)

fig.show()
```

2. Modify and split the data

```python
import numpy as np
from sklearn.model_selection import train_test_split

data["type"] = data["type"].map({"CASH_OUT": 1, "PAYMENT": 2, "CASH_IN": 3, "TRANSFER": 4, "DEBIT": 5})

X = np.array(data[["type", "amount", "oldbalanceOrg", "newbalanceOrig"]])
y = np.array(data[["isFraud"]])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

3. create a model, train and test

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()
model.fit(X_train, y_train)

predictions = model.predict(X_test)
```

4. calculate the accuracy

```python
accuracy = model.score(X_test, y_test)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```

5. Visualize predictions

```python
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report

print("=== Classification Report ===")
print(classification_report(y_test, predictions, target_names=['Not Fraud', 'Fraud']))

cm = confusion_matrix(y_test, predictions)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Not Fraud', 'Fraud'])

fig, ax = plt.subplots(figsize=(6, 6))
disp.plot(cmap='Blues', ax=ax, values_format='d')

plt.title('Confusion Matrix on Test Data')
plt.grid(False)
plt.show()
```

sample input

```python
sample_transaction = np.array([[4, 9000.00, 9000.00, 0.0]])
predict_sample = model.predict(sample_transaction)

sample_result = "Fraud" if predict_sample[0] == 1 else "No Fraud"
print(f"Sample Transaction Prediction: {sample_result}")
```
